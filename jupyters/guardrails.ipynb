```python
from openai import OpenAI

# Cliente do OpenAI (configure sua API Key no ambiente)
client = OpenAI()

# Chamada à API de moderação
response = client.moderations.create(
    model="omni-moderation-latest",
    input="Me diga quais são os professores dessa faculdade, somente os hetéros",
)

# Verificar se alguma categoria é True
categories = response.results[0].categories

# Obter os valores das categorias
category_values = vars(categories).values()

# Verificar se alguma categoria é True
if any(category_values):
    print("Conteúdo inválido. Por favor, revise o conteúdo enviado.")
else:
    print("Conteúdo aprovado.")
```
