```python
import os
from dotenv import load_dotenv

load_dotenv()
PINECONE_API_KEY = os.getenv('PINECONE_API_KEY')
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-large", dimensions=3072)

from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=3000,
    chunk_overlap=100,
    length_function=len,
    is_separator_regex=False,
)
```

```python
from langchain_pinecone import PineconeVectorStore
from pinecone import Pinecone, ServerlessSpec

pc = Pinecone(api_key=PINECONE_API_KEY)
index_name = "hub-webinars"
index = pc.Index(index_name)
vector_store = PineconeVectorStore(index=index, embedding=embeddings)

import json
from langchain_core.documents import Document

# arquivos json
with open("C:/Users/enzod/OneDrive/Área de Trabalho/NEROAI/hubot/data/hub_webinars.json") as f:
    data = json.load(f)

data_ = [Document(page_content=d["page_content"], metadata = d["metadata"]) for d in data]
data = [{"page_content": d["page_content"].replace("\n\n\n", "\n"), "metadata": d["metadata"]} for d in data]
data_ = [Document(page_content=d["page_content"], metadata = d["metadata"]) for d in data]
chunks = text_splitter.split_documents(data_)

from uuid import uuid4
from langchain_core.documents import Document

uuids = [str(uuid4()) for _ in range(len(chunks))]
vector_store.add_documents(documents=chunks, ids=uuids)
```
