```python
import os
from dotenv import load_dotenv

MONGODB_ATLAS_CLUSTER_URI = os.getenv('MONGODB_URI')
```

```python
from pymongo import MongoClient
from datetime import datetime

# Conectar ao MongoDB
client = MongoClient(MONGODB_ATLAS_CLUSTER_URI)
db = client["hubot"]
collection = db["hub-mesas"]

# Dados dos documentos a serem inseridos, incluindo o campo data_update
documentos = [
    {"mesa": 1, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 2, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 3, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 4, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 5, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 6, "status": "disponível", "reservado_por": "", "data_update": ""},
    {"mesa": 7, "status": "disponível", "reservado_por": "", "data_update": ""}
]

# Inserir todos os documentos na coleção de uma vez
try:
    collection.insert_many(documentos)
    print("Documentos inseridos com sucesso!")
except Exception as e:
    print("Erro ao inserir os documentos:", e)
```

```python
from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi


# Create a new client and connect to the server
client = MongoClient(MONGODB_ATLAS_CLUSTER_URI, server_api=ServerApi('1'))

# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```
